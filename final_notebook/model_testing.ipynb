# Model Testing

This notebook loads the saved preprocessor/model and verifies predictions.

```python
import os
import joblib
import numpy as np

from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

```python
MODELS_DIR = "../models"
PROCESSED_DIR = "../dataset/processed"

model = joblib.load(os.path.join(MODELS_DIR, "best_model.pkl"))
preprocessor = joblib.load(os.path.join(MODELS_DIR, "preprocessor.pkl"))

X_test = np.load(os.path.join(PROCESSED_DIR, "X_test.npy"), allow_pickle=True)
y_test = np.load(os.path.join(PROCESSED_DIR, "y_test.npy"), allow_pickle=True)
```

```python
y_pred = model.predict(X_test)

print("Accuracy:", accuracy_score(y_test, y_pred))
print("\nClassification Report:")
print(classification_report(y_test, y_pred))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))
```

```python
# Test one already-preprocessed sample
sample = X_test[[0]]

prediction = model.predict(sample)[0]
probability = model.predict_proba(sample)[0]

print("Prediction:", prediction)
print("Normal probability:", probability[0])
print("Attack probability:", probability[1])

if prediction == 1:
    print("Attack Detected")
else:
    print("Normal Traffic")
```
